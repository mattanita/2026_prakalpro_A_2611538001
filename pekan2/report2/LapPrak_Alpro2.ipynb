Python adalah bahasa pemrograman tingkat tinggi yang bersifat sederhana, dinamis, dan serbaguna (general-purpose). Python pertama kali dikembangkan oleh Guido van Rossum dan dirilis pada tahun 1991. Git adalah sistem kontrol versi (version control system) yang digunakan untuk mencatat dan mengelola perubahan pada kode selama proses pengembangan perangkat lunak. Dengan menggunakan Git, setiap perubahan pada proyek dapat dilacak dan dikelola dengan lebih terstruktur.

### 2.2.5 Program Konstanta_2611531021

```python
# Buat file dengan nama Konstanta_2611538001
# Program ini menggunakan konstanta untuk menghitun luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_8001

from typing import Final
PI: Final = 3.14
print("Pi: %f" % (PI))
jari_8001 = float(input("Masukkan nilai jari-jari: "))
luas_8001 = PI * jari_8001
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_8001,luas_8001))
```

Penjelasan:
