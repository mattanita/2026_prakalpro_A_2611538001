### 1. Python 
 Python is a computer programming language commonly used to build websites, software/applications, automate tasks, and perform data analysis. It is a general-purpose programming language, meaning it can be used to create many different types of programs rather than being limited to a specific problem.
 Because it is versatile and easy to use, it has become one of the most widely used programming languages, especially for beginners.<br>

### 2. Visual Studio Code (VS Code)  
 Visual Studio Code is a free code editor that can run on desktop devices using Linux, Windows, and macOS, and is developed by Microsoft. Visual Studio Code is a powerful yet lightweight software editor.

 Visual Studio Code has several useful features. The following are some of its features:<br>

  a. Basic Editing<br>
  b. IntelliSense<br>
  c. Debugging<br>
  d. Extension Marketplace<br>
  e. GitHub Integration<br>

### 3. Git and GitHub 
Git is a distributed version control system that allows developers to track changes made to source code during software development. With Git, every change made to the code is stored in a repository, making it easier for developers to manage software versions and code change history.<br>
GitHub is a cloud-based service that provides a place to host Git repositories.
