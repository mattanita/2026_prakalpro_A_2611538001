## 2.1.1 Basic Concept of control flow and Selection

<div align="center">

</div>

Control flow is a mechanism in programming languages that determines the order in which instructions or lines of code are executed by the interpreter. In a standard execution process, a program runs sequentially from top to bottom. However, to develop dynamic programs, selection is required to allow the program to choose a particular execution path based on the evaluation of a conditional expression.

Conditional expressions generally produce a Boolean value, which can be either True or False. Based on this value, the program determines whether a specific block of code should be executed. Therefore, control flow and selection structures are important concepts for creating programs that can make decisions according to different conditions and inputs.
<br>

## 2.1.2 Tybe of Conditional Structures in Python
### 1. Single Branching (if)

The if statement is the most basic form of branching structure. The block of code inside the if statement will only be executed when the condition being evaluated returns True. If the condition returns False, the interpreter skips the block of code and continues with the next instruction.
<br>

### 2. Double Branching (if-else)

The if-else structure provides two alternative execution paths. If the condition in the if statement evaluates to True, the code inside the if block is executed. Conversely, if the condition evaluates to False, the program automatically executes the code inside the else block.
<br>

### 3. Multiple Branching (if-elif-else)

The if-elif-else structure is used when a program needs to evaluate multiple conditions. The program checks the conditions sequentially, starting from the if condition and continuing to each elif condition. When one condition evaluates to True, its corresponding block of code is executed, and the remaining conditions are skipped. If none of the conditions are True, the code inside the else block is executed.
<br>

### 4. Nested Branching (Nested IF)

Nested branching refers to a situation in which an if, if-else, or if-elif-else statement is placed inside the code block of another if statement. This structure is used to process multiple levels of conditions where one condition depends logically on another condition.
<br>

### 5. Conditional Expression (Ternary Operator)
Python supports writing a simple conditional statement in a single line of code using a ternary expression. This syntax can be used to simplify the assignment of a variable's value based on a particular condition. The general
From is:
`value_if_true if condition else value_if_false`
The expression returns value_if_true when the condition is True; otherwise, it returns value_if_false.
<br>

### 6. Structural Pattern Matching (match-case)

Starting from Python version 3.10, Python introduced the Structural Pattern Matching feature using the match and case keywords. This structure is similar to the switch-case statement found in other programming languages such as C++ and Java, while providing more flexible pattern-matching capabilities.

The match-case structure evaluates a value or variable against several defined patterns. The underscore character (_) is used in case _: as a wildcard or default case. It is executed when none of the preceding patterns match the given value.
<br>





```python
match variable:
    case value_1:
        # code block if variable == value_1
    case value_2:
        # code block if variable == value_2
    case _:
        # default code block if no pattern matches
```
