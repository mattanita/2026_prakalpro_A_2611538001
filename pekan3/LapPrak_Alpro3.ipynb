<h2><b>1.1 Background</b></h2>

Python is one of the most widely used programming languages in the field of technology. It is known for its simple syntax and is suitable for beginners who are learning programming and algorithms. Therefore, understanding the basic concepts of Python is important for students who are studying programming.

One of the basic concepts in Python is the use of operators. Operators are symbols or keywords used to perform operations on values and variables. Python provides several types of operators, such as arithmetic operators, comparison operators, logical operators, assignment operators, and other operators. These operators are commonly used to perform calculations, compare values, assign data, and create conditions in a program.

Understanding operators is important because they are frequently used when creating Python programs. By learning how different operators work, students can write programs more effectively and understand how Python processes data and expressions.
